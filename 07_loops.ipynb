```python
print("Stop at the first mark below 50:")
for mark in [88,74,45,90]:
    if mark < 50:
        print(f" Found {mark}, stopping.")
        break
    print(f" {mark} is fine.")
print("The loop has ended.")    
```

```python
print("Print only the passing marks:")
for mark in [88,74,45,90]:
    if mark < 50:
        continue
    print(f" {mark} is fine,")
```

```python
# //Accumulation Pattern
marks = [78, 85, 62, 90, 74]
total = 0
for mark in marks:
    total = total + mark
print("Total :", total)
print("Average :", total/len(marks))    
```
