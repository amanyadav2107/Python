How to use numpy to generate random numbers?

```python
#Generate 10 random float
import numpy as np
x=np.random.rand(10) #Old API
print(x)
```

```python
#Generate (3x4) vector of random float 
x=np.random.rand(3,4)
print(x)
```

```python
#Generate 10 random integer in between 5 to 10
x=np.random.randint(5,10,size=10)
print(x)
```

```python
#Generate (3x2) vector to random integer in between 5 to 10
x=np.random.randint(5,11,size=(3,2))
print(x)
```
