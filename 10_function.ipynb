Temperature Conversion Using definition 

```python
def temperature(temp,unit):
    if unit=='C':
        return (temp*9/5)+32
    elif unit=='F':
        return(temp-32)*5/9
    else:
        return None
print(temperature(37,'C'))
    
```

Check Password

```python
def password(pwd):
    if len(pwd)< 8:
        return "Incorrect"
    if not any(char.isdigit() for char in pwd):
        return "Incorrect"
    if not any(char.islower() for char in pwd):
        return"Incorrect"
    if not any(char.isupper() for char in pwd):
        return"Incorrect"
    if not any(char in '&#_!@$' for char in pwd):
        return"Incorrect"
    return"Correct"
print(password("weakpass"))
print(password("&O@Mk123"))       
```
