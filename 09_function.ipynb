```python
def even_odd(num):
    if num %2 ==0:
        print("Even")
    else:
        print("Odd")
even_odd(4)        
```

Default parameters:- 

```python
def great(name ="Ram"):
    print (f"Hello { name}")
great () 
great ("Shyam")  
```

Positional Arguments:- 

```python
def student(name, age):
    print(f"name: {name}")
    print(f"age : {age}")
student("Ram",27) 
student(27,"Ram")   
```

Keyword Argument:- 

```python
def add(*args):
    print(args)
add(10,20,30,40)    
```

Variable length Positional Arguments:- 

```python
def add(*args):
    total =0
    for i in args:
        total = total+i
    return total 
sum = add (10,20,30,40) 
print(f"The total is {sum}") 
```

Variable length keyword argument 

```python
def student(**kwargs):
    print(kwargs)

student(name="Ram", age=27, city="Ghz")  
```
