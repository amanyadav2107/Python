```python
#Fancy indexing returns a copy
import numpy as np
arr = np.array([1,2,3,4])
arr1 = np.array([[1,2,3,4],[5,6,7,8]])
f=arr[[1,2]]
f1=arr1[[0,1],[1,2]]
f[0] =500
print(arr) #unchanged
print(f)
print(f1)
print(f.base)
print(f1.base)
```
