```python
#Spliting the array
x=np.arange(1,13)
np.split(x,4) #4 means equal split in 4 parts 
np.split(x,6) #6 means...................
```

```python
x=np.arange(1,10)
np.split(x,[3,5,6])#  #start spliting from 3rd then from 5th then from 6th
```

```python
x=np.arange (1,10).reshape([3,3])
print(x)
np.hsplit(x,3)
```

```python
x=np.arange (1,11).reshape([2,5])
print(x)
np.hsplit(x,[1,2])
```

```python
#vsplit
x=np.arange (1,11).reshape([5,2])
print(x)
np.vsplit(x,[1,3])
```
