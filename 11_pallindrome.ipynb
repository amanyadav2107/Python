Str Pallindrome check, including both upper and lower case 

```python
def pallindrome(s):
    s=s.lower().replace("_"," ")
    return s==s[::-1]

print(pallindrome("Madam"))
print(pallindrome("Hello"))
```

Factorial

```python
def factorial (num):
    if num == 0:
        return 1
    else:
        return num*factorial(num-1)
print(factorial(6))    
```

```python
def addition(a,b):
    return a+b
addition= lambda a,b: a+b

print(addition(2,3))
print(type(addition))
addition = lambda*args:sum(args)
print(addition(2,3,5,6))
```
